## K-means Clustering
Pada pembahasan kali ini akan diuji 2 metode pemilihan nilai cluster (K) yang terbaik, mendekati distribusi pada label anotator.
1. Metode Elbow
2. Via-Score plot

### Metode Elbow

```python
from sklearn.cluster import KMeans
inertia = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=0)
    kmeans.fit(df_scalling.values)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(20, 10))

sns.lineplot(x=range(1, 11), y=inertia, color='#000087', linewidth = 4)
sns.scatterplot(x=range(1, 11), y=inertia, s=300, color='#800000',  linestyle='--')
plt.title('Elbow Method - Finding K value Clusters')
plt.xlabel('The Amount of K')
plt.ylabel('Intertia / WSS')
```

```python
#Dari hasil diatas elbow nya dipilih pada angka 3,
# karena titik setelah lekukan tajam sebelum grafik mulai mendatar ada di K=3
from sklearn.cluster import KMeans
kmeans_elbow = KMeans(n_clusters=3, random_state=0)
kmeans_elbow.fit(df_scalling.values)
```

```python
#taruh hasil k-means elbow method ke df dengan nama kolom cluster_elbow
df['cluster_elbow'] = kmeans_elbow.labels_
```

```python
df
```

```python
fig, ax = plt.subplots(figsize=(15,10))
sns.scatterplot(data=df, x='fixed acidity', y='alcohol', hue='cluster_elbow')
```

### Bandingkan hasil dengan label anotator

```python
fig, ax = plt.subplots(figsize=(15,10))
sns.scatterplot(data=df, x='fixed acidity', y='alcohol', hue='quality')
```

### Hasil diatas ketika menggunakan elbow ialah cukup optimal. Dengan kondisi :
1. Cluster 0 -> wine dengan fixed acidity tinggi dan alcohol tinggi, cenderung kualitas lebih baik
2. Cluster 1 -> wine dengan fixed acidity sedang dan alcohol rendah, cenderung kualitas menengah ke bawah
3. Cluster 2 -> wine dengan fixed acidity rendah dan alcohol sedang, kualitas menengah

### 2. Via Score Plot

```python
# Via Score Plot
from yellowbrick.cluster import KElbowVisualizer
k_means_via = KMeans()
# k is range of number of clusters.
visualizer = KElbowVisualizer(k_means_via, k=(2,11), metric='silhouette', timings= True, force_model=True)
visualizer.fit(df_scalling.values)        # Fit the data to the visualizer
plt.title('Distortion Via-Score Plot for K-Means Clustering')
plt.show()
```

```python
# Dari hasil diatas K-nya dipilih pada angka 2,
# karena nilai silhouette score tertinggi ada di K=2
from sklearn.cluster import KMeans
kmeans_via = KMeans(n_clusters=2, random_state=0)
kmeans_via.fit(df_scalling.values)
```

```python
#taruh hasil k-means via score method ke df dengan nama kolom cluster_via
df['cluster_via'] = kmeans_via.labels_
```
